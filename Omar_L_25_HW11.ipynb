```python
class ChargedParticle:
    def __init__(self, mass, px, py, pz, vx, vy, vz, charge):
        """
        Initialize a charged particle with mass, position, velocity, and charge
        """
        self.mass = mass
        # Position coordinates
        self.px = px
        self.py = py
        self.pz = pz
        # Velocity components
        self.vx = vx
        self.vy = vy
        self.vz = vz
        # Charge of the particle
        self.charge = charge
    
    # This method is needed for the simulation to work
    def move(self, dt):
        """
        Update the position of the particle based on its velocity
        """
        self.px += self.vx * dt
        self.py += self.vy * dt
        self.pz += self.vz * dt
    
    def find_separation(self, other):
        """
        Calculate the separation distance between this particle and another
        """
        dx = self.px - other.px
        dy = self.py - other.py
        dz = self.pz - other.pz
        return (dx**2 + dy**2 + dz**2)**0.5
    
    def same_place(self, other):
        """
        Check if two particles are in the same place
        """
        separation = self.find_separation(other)
        # For demonstration, just print the separation
        print(f"Separation: {separation}")
        return separation < 0.1  # Assuming particles are "same place" if within 0.1 units


#7.
p1 = ChargedParticle(10, -5, -5, -5, 1, 1, 1, 0.5)
p2 = ChargedParticle(10,  5,  5,  5, -1, -1, -1, -0.5)

dt = 0.05

for step in range(300):
    p1.move(dt)
    p2.move(dt)
    p1.same_place(p2)
    print("step:", step, " P1 position =", (p1.px, p1.py, p1.pz), " velocity =", (p1.vx, p1.vy, p1.vz),
          " | P2 position =", (p2.px, p2.py, p2.pz), " velocity =", (p2.vx, p2.vy, p2.vz))


#8.
p1 = ChargedParticle(10, -5, -5, -5, 2, 2, 2, 0.5)
p2 = ChargedParticle(10,  5,  5,  5, -1, -1, -1, 2.0)

dt = 0.05

for step in range(300):
    p1.move(dt)
    p2.move(dt)
    p1.same_place(p2)
    print("step:", step, " P1 position =", (p1.px, p1.py, p1.pz), " velocity =", (p1.vx, p1.vy, p1.vz),
          " | P2 position =", (p2.px, p2.py, p2.pz), " velocity =", (p2.vx, p2.vy, p2.vz))
```
